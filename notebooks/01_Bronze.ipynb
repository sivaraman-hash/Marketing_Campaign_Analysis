# 01 Bronze: Raw Data Ingestion
- **Source:** Kaggle Customer Personality Analysis (marketing_campaign.csv, tab-separated)
- **What it does:** Loads the raw file from the Volume with no changes
- **Output table:** marketing.default.bronze_campaign


```python
# Bronze: load raw data exactly as it is
file_path = "/Volumes/marketing/default/raw_data/marketing_campaign.csv"

df_raw = (spark.read
    .option("header", True)
    .option("inferSchema", True)
    .option("sep", "\t")
    .csv(file_path))

df_raw.write.mode("overwrite").saveAsTable("marketing.default.bronze_campaign")

display(df_raw.limit(10))
print("Rows:", df_raw.count())
```
