# 04 Quality Check
Validates the Silver table before Gold runs: row count, no null income, valid marital status, no negative spend.
If any check fails, the job stops and Gold does not run.

```python
df = spark.table("marketing.default.silver_campaign")

total = df.count()
null_income = df.filter("Income IS NULL").count()
bad_marital = df.filter("Marital_Status NOT IN ('Partnered','Single')").count()
neg_spend = df.filter("TotalSpend < 0").count()

print("Rows:", total, "| Null income:", null_income,
      "| Bad marital:", bad_marital, "| Negative spend:", neg_spend)

assert total > 1000, "Too few rows in Silver"
assert null_income == 0, "Null income found"
assert bad_marital == 0, "Unexpected marital status"
assert neg_spend == 0, "Negative spend found"
print("All quality checks passed")
```
